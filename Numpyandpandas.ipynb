```python
import numpy as np
import pandas as pd

# NUMPY ARRAY OPERATIONS

print("NUMPY ARRAY OPERATIONS\n")

# 1.Create a 1D NumPy Array
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
print("1D Array (temperatures_w1):", temperatures_w1)

# 2.Inspection and Properties
print("\nShape:", temperatures_w1.shape)
print("Data Type:", temperatures_w1.dtype)
print("Number of Elements:", temperatures_w1.size)

# 3.Array Operations
temperatures_w1_f = (temperatures_w1 * 9 / 5) + 32
print("\nTemperatures in Fahrenheit:", temperatures_w1_f)
print("Maximum Temperature:", np.max(temperatures_w1))
print("Minimum Temperature:", np.min(temperatures_w1))
print("Mean Temperature:", np.mean(temperatures_w1))

# 4.Array Slicing and Indexing
print("\nFirst three days:", temperatures_w1[:3])
print("Weekend (last two days):", temperatures_w1[-2:])
print("Middle three days:", temperatures_w1[2:5])

# 5.Create a 2D Array
temperatures = np.array([[22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],  
        [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1],])

# 6.Inspect and Slice the 2D Array
print("\n2D Array Shape:", temperatures.shape)
print("2D Array Data Type:", temperatures.dtype)
print("2D Array Total Elements:", temperatures.size)

print("\nWeek 1 Temperatures:", temperatures[0])
print("Week 2 Temperatures:", temperatures[1])
print("Weekends (both weeks):")
print("Week 1 Weekend:", temperatures[0, -2:])
print("Week 2 Weekend:", temperatures[1, -2:])

# PANDAS SERIES

print("PANDAS SERIES\n")

# 1. Creating Pandas Series
marks_data = [95, 92, 89, 85, 80]
indices = ["Rank1", "Rank2", "Rank3", "Rank4", "Rank5"]
marks = pd.Series(marks_data, index=indices)
print("Marks Series:")
print(marks)

# 2. Indexing and Slicing
print("\nMark of 1st rank student (integer index):", marks.iloc[0])
print("\nMarks of top 3 ranks (.loc):")
print(marks.loc[["Rank1", "Rank2", "Rank3"]])
print("\nMark of 3rd rank student (.iloc):", marks.iloc[2])

print("\nRanks where marks > 90:")
print(marks[marks > 90])

# 3. Manipulating Series
marks["Rank1"] = 100
print("\nAfter modifying 1st rank to 100:")
print(marks)

marks = marks.drop("Rank5")
print("\nAfter removing the last rank:")
print(marks)

cgpa = marks / 10
print("\nCGPA:")
print(cgpa)


#PANDAS DATAFRAME

print("\n\nPANDAS DATAFRAME\n")



#1Creating DataFrame
data = {
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'ProductCategory': ['Electronics', 'Clothing', 'Electronics', 'Furniture',
                        'Clothing', 'Electronics', 'Furniture', 'Clothing',
                        'Furniture', 'Electronics'],
    'Region': ['North', 'South', 'North', 'East', 'West',
               'North', 'East', 'West', 'South', 'North'],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
}

transactions = pd.DataFrame(data)

#2Data Exploration
print(transactions)
print(transactions.head())
print(transactions.tail())
print(transactions.shape)
print(transactions.columns)
print(transactions.info())

print(transactions[['ProductCategory', 'Amount']])
print(transactions.iloc[:, -3:])

print(transactions[(transactions['Region'] == 'North') &
                   (transactions['Amount'] > 200)])

print(transactions['ProductCategory'].value_counts())
print(transactions['Region'].unique())

print(transactions.groupby('Region')['Amount'].mean())

#3.Manipulating DataFrame

#Modify Amount
transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165

#Add Discount column
transactions['Discount'] = transactions['Amount'] * 0.10

#Remove TransactionID 109
transactions = transactions[transactions['TransactionID'] != 109]

#Delete Discount column
transactions.drop('Discount', axis=1, inplace=True)

#Display final DataFrame
print(transactions)
```
